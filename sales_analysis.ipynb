# Phase 2 — Data cleaning and preparation


We will clean a working copy of the data, keeping raw_df unchanged. The steps are to trim whitespace, standardize common missing-value markers, normalize category labels, inspect malformed numeric values, convert quantity and price, remove exact duplicate rows, and fill missing numeric values using the median for the same product where possible.


Filling by product assumes that values for the same product are more comparable than values across different products. Missing values that cannot be filled from a product group will remain missing rather than being assigned an unsupported value. Repeated order_id values will not be removed automatically.

````
# Trim whitespace from text columns.
text_columns = df.select_dtypes(include=["object", "string"]).columns

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

print("Whitespace trimmed from text columns.")
````

````
# Drop duplicate rows, keeping the first occurrence.
initial_row_count = df.shape[0]
df = df.drop_duplicates(keep="first")
final_row_count = df.shape[0]
print(f"Dropped {initial_row_count - final_row_count:,} duplicate rows. Remaining rows: {final_row_count:,}.")
````

````
for column in df.columns:
    df[column] = df[column].astype("string").str.strip().replace({"": pd.NA, "NA": pd.NA, "N/A": pd.NA, "null": pd.NA, "NULL": pd.NA})
    df[column] = df[column].str.title()
````

````
df['customer_id'] = df['customer_id'].str.upper()
````

````
import pandas as pd

df["qty_original"] = df["qty"]

qty_text = (
    df["qty"]
    .astype("string")
    .str.strip()
    .str.casefold()
    .str.replace(r"\s*(?:pcs?|units?)\s*$", "", regex=True)
)

# Handle number words present in the data.
qty_text = qty_text.replace({"two": "2", "five": "5"})

qty_numeric = pd.to_numeric(qty_text, errors="coerce")

# Round fractional quantities to whole numbers, then convert to nullable integers.
df["qty"] = qty_numeric.round().astype("Int64")
````

````
print(df.price.isna().sum())
````

````
df["price_original"] = df["price"]

price_text = (
    df["price"]
    .astype("string")
    .str.strip()
    .str.casefold()
    .str.replace(r"^[n₦]\s*", "", regex=True)
    .str.replace(",", "", regex=False)
    .str.replace(r"\s+", "", regex=True)
)

# Identify and remove a trailing "k", then multiply those values by 1,000.
in_thousands = price_text.str.endswith("k", na=False)
price_text = price_text.str.replace(r"k$", "", regex=True)

price_numeric = pd.to_numeric(price_text, errors="coerce")
price_numeric.loc[in_thousands] *= 1000

df["price"] = price_numeric.round().astype("Int64")
````

````
df.head(20)
````

````
import pandas as pd

df["order_date_original"] = df["order_date"]

date_formats = [
    "%Y-%m-%d",       # 2025-05-20
    "%Y/%m/%d",       # 2025/05/20
    "%Y|%m|%d",       # 2025|05|20
    "%b %d, %Y",      # Apr 22, 2025 or Oct 22, 2025
    "%B %d, %Y",      # April 22, 2025 or October 22, 2025
    "%d %b %Y",       # 22 Oct 2025
    "%d %B %Y",       # 22 October 2025
    "%d/%m/%Y",       # 26/07/2025; day-first, as indicated by the data
    "%m/%d/%Y",       # Used only if day-first parsing fails
    "%m-%d-%Y",       # 07-23-2025
    "%d-%m-%Y",
]

def parse_order_date(value):
    if pd.isna(value):
        return pd.NaT

    text = str(value).strip()

    if not text or text.casefold() in {
        "na", "n/a", "not recorded", "invalid", "unknown"
    }:
        return pd.NaT

    for date_format in date_formats:
        parsed = pd.to_datetime(text, format=date_format, errors="coerce")
        if not pd.isna(parsed):
            return parsed

    return pd.NaT

df["order_date"] = pd.to_datetime(df["order_date"].map(parse_order_date))
````

````
parse_failed = (
    df["order_date_original"].notna()
    & df["order_date"].isna()
)

print("Missing or unparsed dates:", df["order_date"].isna().sum())
print("\nOriginal values that could not be parsed:")
display(df.loc[parse_failed, ["order_date_original"]].drop_duplicates())

print("\nConverted date type:")
print(df["order_date"].dtype)
````

````
# Normalize payment method labels.
payment = df["payment_method"].astype("string").str.strip()
df.loc[payment.str.casefold().eq("bank transfer"), "payment_method"] = "Transfer"

# Normalize status labels.
status = df["status"].astype("string").str.strip()
df.loc[status.str.casefold().eq("complete"), "status"] = "Completed"
df.loc[status.str.casefold().eq("return"), "status"] = "Returned"
````

````
import pandas as pd

# Remove non-alphanumeric characters from customer IDs and keep uppercase.
# Missing IDs remain missing.
df["customer_id"] = (
    df["customer_id"]
    .astype("string")
    .str.replace(r"[^A-Za-z0-9]+", "", regex=True)
    .str.upper()
    .replace("", pd.NA)
)

# Separate rows with and without an order ID so missing IDs aren't all merged.
has_order_id = df["order_id"].notna()
with_id = df.loc[has_order_id].copy()
without_id = df.loc[~has_order_id].copy()

# Prefer the least-missing row for each order ID; stable sorting preserves
# the original row order when rows have the same number of missing values.
with_id["_missing_count"] = with_id.isna().sum(axis=1)
with_id = with_id.sort_values("_missing_count", kind="stable")
with_id = with_id.drop_duplicates(subset="order_id", keep="first")
with_id = with_id.drop(columns="_missing_count")

# Keep rows with missing order IDs, then sort IDs highest to lowest.
df = pd.concat([with_id, without_id], ignore_index=True)
df["_order_id_sort"] = pd.to_numeric(df["order_id"], errors="coerce")
df = (
    df.sort_values("_order_id_sort", ascending=False, na_position="last", kind="stable")
      .drop(columns="_order_id_sort")
      .reset_index(drop=True)
)
````

````
# Preserve which values were actually missing before estimating anything.
df["qty_was_missing"] = df["qty"].isna()
df["price_was_missing"] = df["price"].isna()

# Fill these categories for grouping/display; this does not invent a customer identity.
for column in ["product", "region"]:
    df[column] = df[column].astype("string").str.strip().fillna("Unknown")
````

````
df.head()
````

````
# Fill missing quantity with the average quantity for that product.
df["qty"] = df["qty"].fillna(
    df.groupby("product")["qty"].transform("mean").round().astype("Int64")
)

# Fill missing price with the average price for that product and region.
df["price"] = df["price"].fillna(
    df.groupby(["product", "region"])["price"]
      .transform("mean")
      .round()
      .astype("Int64")
)
````

````
# Remove raw numeric text columns; keep qty_was_missing and price_was_missing.
df = df.drop(columns=["price_original", "qty_original"], errors="ignore")

# Keep the full dataset and its parsed order_date column as the cleaned snapshot.
# Keep order_date_original if you want to retain the source date text for auditing.
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")

# Use only dated rows for time-based analysis.
dated_sales = df.loc[df["order_date"].notna()].copy()

print("Rows in full cleaned dataset:", len(df))
print("Rows available for date-based analysis:", len(dated_sales))
print("Rows with missing dates:", df["order_date"].isna().sum())
````

## Phase 2 — Data Cleaning and Preparation

Missing quantities and prices were filled using group averages, while flags were retained to identify originally missing values. Duplicate order IDs were handled by keeping the most complete row. Rows with missing dates were preserved in the full dataset and excluded only from date-based analysis.
